# 2. Get the Data 
Define how you recieved the data (provided, gathered..)

```python
from PIL import Image
from pillow_heif import register_heif_opener
import pandas as pd
import pyprog
import time
def load(folder):
    bar=pyprog.ProgressBar("[","]",500)
    bar.update()
    register_heif_opener()
    listo=[]
    i=0
    stt=time.time()
    for dirname, _, filenames in os.walk(folder):
        for filename in filenames:
            if len(filename.split(".")[0].lower())>1:
                label="{"
                print(dirname)
            else:
                label=filename.split(".")[0].lower()
            img=np.asarray(Image.open(os.path.join(dirname, filename)).convert("RGB"))
            h,w=img.shape[:2]
            if h<220 or w<220:
                print("waaah")
            img_data=img[h//2-100:h//2+100,w//2-100:w//2+100]
            img_buf=[]
            for t in range(20):
                img_data=img[(h-10+t)//2-100:(h-10+t)//2+100,(w-10+t)//2-100:(w-10+t)//2+100]
                if img_data.shape[1]==200 and img_data.shape[0]==200:
                    listo.append([img_data,int(ord(label))-97])
                else:
                    print("exception:",img_data.shape[:2])
            #print(filename.split(".")[0])
            bar.set_stat(i)
            i+=1
            bar.update()
    df=pd.DataFrame(data=listo,columns=["data","labels"])
    bar.end()
    print("Done!")
    for i in range(df.shape[0]):
        df.iloc[i].to_csv(f"./data_proc/{i}",header=False)
    print(f"elapsed time:{time.time()-stt:.2f} seconds")
    return df
data=load("./Data")
data["labels"].value_counts()
```

# 8. Launch the Model System
Define your production run code, This should be self susficent and require only your model pramaters 


```python
import torch_esy as tor
import numpy as np
from PIL import Image
from pillow_heif import register_heif_opener
import pandas as pd
import pyprog
import time
import torch
from torch import nn
dict_={"a":1,"b":2,"c":3,"d":4,"e":5,"f":6,"g":7,"h":8,"i":9,"j":10,"k":11,"l":12,"m":13,"n":14,"o":15,"p":16,"q":17,"r":18,"s":19,"t":20,"u":21,"v":22,"w":23,"x":24,"y":25,"z":26}
dict_ = {v: k for k, v in dict_.items()}
def inference(img_path):
    register_heif_opener()
    img=np.asarray(Image.open(img_path).convert("RGB"))
    model=tor.model(0,load=True)
    model.eval()
    probs=model(torch.tensor(np.asarray(img)[np.asarray(img).shape[0]//2-100:np.asarray(img).shape[0]//2+100,np.asarray(img).shape[1]//2-100:np.asarray(img).shape[1]//2+100],dtype=torch.float32).permute(2,0,1).unsqueeze(0))
    return dict_[int(torch.argmax(probs))]
inference("./Data/S26/z.HEIC")
```
